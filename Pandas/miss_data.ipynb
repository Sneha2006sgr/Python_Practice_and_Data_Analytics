```python
import numpy as np
import pandas as pd
```

#### Finding Missing Data

```python
data={
    'A':[1,2, np.nan,4,5],
    'B':[1,2,3,4, 5],
    'C':[1,2,3,np.nan,np.nan],
    'D':[np.nan,2,np.nan,np.nan,5]
}
```

```python
df=pd.DataFrame(data)
df
# nan is Not A Number
```

```python
df.isnull()
# Check if data is null or not
```

```python
df.isnull().sum()
# Count the nulls data
```

```python
df.isna()
```

```python
df.isna().sum()
```

```python
df.isna().any()
# Checking which columns have the null values
```

Removing Missing data

```python
df.dropna()
# Removes all rows having single null data

# Dropna works row wise
```

```python
df
```

```python
df.dropna(thresh=3)
#  threshhold = have same number of non null values
```

```python
df.dropna(thresh=4)
```

#### Filling the Missing Data

```python
df
```

```python
df.fillna(0)

# fillna works column wise
```

```python
values={
    'A':0,
    'B':100,
    'C':200,
    'D':300
}

df.fillna(values)
```

```python
df.fillna(df.mean())
```
