# Pivot Tables


Values should be always numeric else it will give error as it performs maths calculations


General purpose spreadsheet tool that aggregates, summarizes and pivots any flat list of records using count, sums or average

It is used when you want to use your values your indexes which is not already defined

Used to create heat maps

```python
import numpy as np
import pandas as pd
```

```python
data={
    'Date': pd.date_range('2026-01-01',periods=20),
    'Product': ['A', 'B', 'C', 'D']*5, 
    'Region': ['East', 'West', 'North', 'South', 'East', 'West', 'North', 'South', 'East', 'West', 'North', 'South', 'East', 'West', 'North', 'South', 'East', 'West', 'North', 'South'],
    'Sales': np.random.randint(100, 1000, 20),
    'Units': np.random.randint(10, 100, 20),
    'Rep': ['Rahul', 'Harshal', 'Amit', 'Shubham', 'Rahul', 'Harshal', 'Amit', 'Shubham', 'Rahul', 'Harshal', 'Amit', 'Shubham', 'Rahul', 'Harshal', 'Amit', 'Shubham', 'Rahul', 'Harshal', 'Amit', 'Shubham']
}

df=pd.DataFrame(data)
df['Month']=df['Date'].dt.month_name()
df['Quarter']='Q'+ df['Date'].dt.quarter.astype(str)
```

```python
df
```

```python
#  converting A, B, C, D as columns and East, West, North and South as Regions in columns

pd.pivot_table(df,values='Sales', index='Region', columns='Product')
```

```python
pvot=pd.pivot_table(df,values='Sales', index='Region', columns='Product',aggfunc='median')
```

```python
pivot2=pd.pivot_table(df,values=['Sales','Units'], index='Region', columns='Product')
```
