```python
import numpy as np
import pandas as pd
```

### Merging Joining and Concatenation

```python
employees=pd.DataFrame({
    'emp_id':[1, 2, 3, 4, 5],
    'emp_name':['Harshal', 'Kishor', 'Sangam', 'Anand', 'Pooja'],
    'emp_dept':['HR','IT','finance','IT','HR']
})

salaries=pd.DataFrame({
    'emp_id':[1, 2, 3, 6, 7],
    'salary':[50000, 75000, 62000, 70000, 80000],
    'bonus':[4000, 9000, 7000, 7000, 12000]
})
```

```python
employees
```

```python
salaries
```

```python
pd.merge(employees,salaries,on='emp_id')
```

```python
pd.merge(employees,salaries,on='emp_id',how='inner')
```

```python
pd.merge(employees,salaries,on='emp_id',how='outer')
```

```python
pd.merge(employees,salaries,on='emp_id',how='left')
```

```python
pd.merge(employees,salaries,on='emp_id',how='right')
```

### Concatenation of dataframes

```python
df1=pd.DataFrame({
    'A':['A0', 'A1', 'A2'],
    'B':['B0', 'B1', 'B2'],
    'C':['C0', 'C1', 'C2']
})

df2=pd.DataFrame({
    'A':['A3', 'A4', 'A5'],
    'B':['B3', 'B4', 'B5'],
    'C':['C3', 'C4', 'C5']
})
```

```python
df1
```

```python
df2
```

```python
pd.concat([df1,df2])
```

```python
pd.concat([df2,df1])
```

```python
pd.concat([df1,df2],axis=1)
```

**Joining two dataframes**

```python
df1=pd.DataFrame({
    'name':['Prerana', 'Rucheeta', 'Palak', 'Sid']
},index=[1,2,3,4])

df2=pd.DataFrame({
    'score':[89, 90, 75, 98]
}, index=[2,3,4,5])
```

```python
df1
```

```python
df2
```

```python
df1.join(df2)
```

```python
df2.join(df1)
```

```python
df1.join(df2,how='outer')
```
