Feature Extraction

Pandas support below files
1. csv => pd.read_csv('file.csv')
2. excel => pd.read_excel('file.xlsx')
3. json => pd.read_json('file.json')
4. text => pd.read_csv('file.txt', delimiter=' ')
5. tsv => pd.read_tsv('file.tsv', sep='/t')
6. html tables => pd.html('file.html')
7. sql databases => pd.read_sql(query,connection)
8. parquet => pd.read_parquet('file.parquet')
9. feather => pd.read_feather('file.feather')
10. pickle => pd.read_pickle('file.pkl')
11. hdf5 (Hierarchial data format) => pd.read_hdf('file.h5')
12. orc (Optimizezd Row Columnar) => pandas.read_orc()
13. google sheets => use gpread, pandas+API
