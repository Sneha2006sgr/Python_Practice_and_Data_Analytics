# Group By

done categorically

Qualitative Data:
- Nominal Data: Categorical ( movie genre, eye color, brands)
- Ordinal Data: Ranking of Data ( poor, good, bad)

Quantitative Data:
- Continuous Data (1.23, 1.24, 1.25)
- Discrete Data (1, 2, 3, 4)

```python
import numpy as np
import pandas as pd
```

```python
data={
    'Category':['A', 'B', 'A', 'B', 'A', 'B', 'A', 'B'],
    'Store':['S1', 'S1', 'S2', 'S2', 'S1', 'S2', 'S2', 'S1'],
    'Sales': [100, 200, 150, 250, 120, 180, 200, 300],
    'Quantity':[10, 15, 12, 18, 8, 20, 15, 25],
    'Date':pd.date_range('2023-01-01', periods=8)
}
```

```python
df=pd.DataFrame(data)
```

```python
df
```

```python
# Group by Category and calculate the sum of sales
df.groupby('Category')
```

```python
d2=df.groupby('Category')
d2
for i, j in d2:
    print(i)
    print(j)
```

```python
obj=df.groupby('Category')['Sales'].sum()
obj
```

```python
# Group by Store and calculate the sum of Sales
obj2=df.groupby('Store')['Sales'].sum()
obj2
```

```python
#  Group by multiple columns
#  Group by Category and stores
obj1=df.groupby(['Category','Store'])['Sales'].sum()
obj1
```
