# Data Frames

Combining mulitple Series called as Dataframes

```python
import numpy as np
import pandas as pd
```

### Creating a DataFrames

```python
data={
    'Name':['Neha','Rahul','Sachin','Supriya'],
    'Age':[21,30 ,29, 42],
    'City':['Mumbai','Lucknow','Varanasi','Delhi'],
    'Salary':[40000,70000,64000,85000]
}
```

```python
df=pd.DataFrame(data)
df
```

```python
list=[
    ['Jigar',28,'Bangalore',65000],
    ['Anamika', 30,'Karnataka', 70000],
    ['Himanshu', 24,'Pune',45000],
    ['Harshada',45,'Kolkata',85000]
]

cols=['Name','Age','City','Salary']
df2=pd.DataFrame(list,columns=cols)
```

```python
df2
```

### Selection and Indexing of Columns

```python
df2['Name']
```

```python
df2['City']
```

```python
#  To select multiple columns give list of columns as input
df2[['Name','Age','City']]
```

### Creating a new column

```python
# To add a column
df2['Designation']=['Data Analyst', 'Data Scientist','Business Analyst', 'Quality Analyst']
df2
```

### Removing Columns

```python
# Removing a Column
# df2.drop('Designation',axis=0)
df2.drop('Designation',axis=1)

# Axis 0 is Horizontal to access Rows
# Axis 1 is Vertical to access Columns

#  Never removes from main data just you can view the snapshot how the data will look after removing it
```

```python
df2
```

```python
# to remove specific one
df2.drop('Designation', axis=1, inplace=True)

# inplace states that the data is to be removed permanently (Boolean Value to be given by default it is false)
```

```python
df2
#  Now the Designation is permanently removed
```

```python
df2.drop(['City','Age'],axis=1)
```

```python
df2.drop(0,axis=0)
# Drops first row
```
