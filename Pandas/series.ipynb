# Pandas Series

A series is a one-dimensional labeled aaray capable of holding any data type. The axis labels are collectively called the index

# Creating a series

There are multiple ways to create a pandas series

```python
import numpy as np
import pandas as pd
```

```python
labels=['a', 'b', 'c']
list=[10,20,30]
arr=np.array([10,20,30])
dict={'a':10,'b':20,'c':30}
```

```python
pd.Series(list)
```

```python
# index points are labels and data is in series
# you can manually update the index
pd.Series(list,labels)
```

```python
pd.Series(arr,labels)

# 2-d Arrays will throw errors are series are alwayd one dimensional
```

```python
pd.Series(dict)

# Keys are treated as index
```
