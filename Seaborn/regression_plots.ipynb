# Regression plots

# lmplot()

regression plot is used in machine learning algorithm

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

As we predict in Machine learning 
the prediction is at Y axis
the plot is scatter plot the best fit line is drawn which is called completely as regression plot

```python
tip=sns.load_dataset('tips')
```

```python
sns.lmplot(x='total_bill',y='tip', data=tip)
```

The above is linear regression algorithm


```python
sns.lmplot(x='total_bill',y='tip', data=tip,hue='sex')
```

```python
sns.lmplot(x='total_bill',y='tip', data=tip,hue='sex',palette='coolwarm')
```

```python
sns.lmplot(x='total_bill',y='tip', data=tip,hue='sex',palette='rainbow',markers=['o','v'])
```

```python
sns.lmplot(x='total_bill',y='tip', data=tip,hue='sex',palette='rainbow',markers=['o','v'],scatter_kws={'s':70})
```
