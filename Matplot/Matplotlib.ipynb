```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

**Matplotlib Basics**

```python
x=np.linspace(0,5,11)
x
```

```python
y=x**2
y
```

```python
# Creating line graph

plt.plot(x,y)
plt.title("Sample")
plt.xlabel("Number")
plt.ylabel("Squares")
```

**Subplots in Matplotlib**

```python
plt.subplot(2,2,1)
plt.plot(x,y)
plt.subplot(2,2,4)
plt.plot(x,y)
plt.subplot(2,2,3)
plt.plot(y,x)
plt.subplot(2,2,2)
plt.plot(x**2,x**2)
```

**Object Oriented Way**

Matplotlib allows the aspect ratio, DPI, figure size to be specified when the figure object is created. You can use the figsize and dpi  keyword arguments

figsize is a tuple of th ewidht and height of the figure in inches dpi is the dots-per-inch(pixel per inch)

```python
fig=plt.figure(figsize=(6,8),dpi=(100))

fig
```

```python
#  Left, bottom, width, height
#  parameters to be between 0 and 1


axis1= fig.add_axes([0.1,0.1,0.8,0.8])
fig
```

```python
axis1.plot(x,y)
axis2=fig.add_axes([0.1,0.3,0.8,0.8])
axis2.plot(x,y)
fig
```
