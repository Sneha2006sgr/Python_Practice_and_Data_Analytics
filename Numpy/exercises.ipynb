### Valid Sudoku


```python
import numpy as np
```

```python
sudoku=np.array([
    [5, 3, 4, 6, 7, 8, 9, 1, 2],
    [6, 7, 2, 1, 9, 5, 3, 4, 8],
    [1, 9, 8, 3, 4, 2, 5, 6, 7],

    [8, 5, 9, 7, 6, 1, 4, 2, 3],
    [4, 2, 6, 8, 5, 3, 7, 9, 1],
    [7, 1, 3, 9, 2, 4, 8, 5, 6],
    
    [9, 6, 1, 5, 3, 7, 2, 8, 4],
    [2, 8, 7, 4, 1, 9, 6, 3, 5],
    [3, 4, 5, 2, 8, 6, 1, 7, 9]
])
```

```python
b= np.sum(sudoku,axis=1)
b == 45
```

```python
for i in b:
    if i != 45:
        print("invalid Sudoku")
        break
else:
    print("Valid Sudoku for rows")
```

```python
c= np. sum(sudoku,axis=0)
for i in c:
    if i != 45:
        print("Not a valid Sudoku")
        break
else:
    print("Valid sudoku")
```

```python
#  checking if 3 x 3 matrix haves value sum of 45 it it is the sudoku is valid else it is not
#  to access the 3 x 3 matrix will be accessed through slicing
a=sudoku[0:3,0:3]
a
b=a.sum()
b
```

```python
if a.sum()!=45:
    print ("Not a valid sudoku")
    
else:
    print("It is a valid sudoku")
```

```python
for i in range(0,9,3):
    for j in range(0,9,3):
        n=sudoku[i:i+3, j:j+3]
        print(n.sum())
        if n.sum()!=45:
            print("Not a valid Sudoku")
            break
        else:
            print("It is a valid sudoku")
```
