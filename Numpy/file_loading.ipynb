# File Loading using Numpy

### Uploading values in csv file and creating one if not exists

Syntax : numpy.load(file, mmap_mode=None, allow_pickle=True, fix_imports=True, encoding='ASCII')

Parameters:  
 file  : : file-like object, string, or pathlib.Path.The file to read. File-like objects must support the seek() and read() methods. 
 mmap_mode  : If not None, then memory-map the file, using the given mode (see numpy.memmap for a detailed 
  description of the modes). 
 allow_pickle  : Allow loading pickled object arrays stored in npy files.  
 fix_imports  : Only useful when loading Python 2 generated pickled files on Python 3,which includes npy/npz files containing object arrays. 
 encoding  : Only useful when loading Python 2 generated pickled files in Python 3, which includes npy/npz files containing object arrays.

Returns  : Data stored in the file. For .npz files, the returned instance of NpzFile class must be closed to avoid leaking file descriptors.

Numpy has already prebuilt path library so no need to import path library

```python
import numpy as np
```

```python
# Sample student marks
marks = [78, 85, 92, 67, 88, 56, 95, 73]

# Save data into a text file
np.savetxt("marks.txt", marks, fmt="%d")

# fmt is a format specifier specifiying that dta is of decimal 2^10 in format (Natural numbers with no floating points)
# Save data into a CSV file
np.savetxt("marks.csv", marks, delimiter=",", fmt="%d")

# Delimiter could be any special character you need tab, comma, slash or anything else you need|
print("Files created successfully!")
```

```python
data = np.loadtxt("marks.txt")

print(data)
print(type(data))
print(data.dtype)
```

```python
# Load comma-separated numeric data
data = np.loadtxt("marks.csv", delimiter=",")

print(data)
print("Total marks:", np.sum(data))
print("Average:", np.mean(data))
```

```python
# Create a dataset with multiple columns
student_data = np.array([
    [101, 78, 85],
    [102, 92, 88],
    [103, 67, 72],
    [104, 56, 64]
])

# Save the dataset
np.savetxt(
    "students.csv",
    student_data,
    delimiter=",",
    fmt="%d",
    header="ID,Python,SQL",
    comments=""
)

# Load the dataset while skipping the header
data = np.loadtxt(
    "students.csv",
    delimiter=",",
    skiprows=1
)

print(data)
```
