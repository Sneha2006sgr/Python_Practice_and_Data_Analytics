# Numpy Arrays
1. Looks as same as List
2. Have same data type while list have multiple data type
3. Faster than List (convert array to Compiler level) while list are slower being running on interpreter
4. Memory efficiency is higher than list (less space higher performance)
5. Vector Operations are easy and fast while list needs loops (think of performing calculations on all elements of list)

# Random Generation Function 
numpy automatically installs random library

```python
np.random.rand(10)
# Normalization range 0 to 1
```

```python
np.random.randn(10)
# Standardization range -3 to 3
```

```python
np.random.randint(10,30,10)
# random list between 10 to 30 having 10 elements
```

```python
np.random.randint(7)
# random number generating between 0 to 7
```
