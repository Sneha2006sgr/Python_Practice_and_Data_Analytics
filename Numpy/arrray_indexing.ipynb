# NUMPY INDEXING AND SLICING OF VECTORS

```python
arr=np.arange(11,21)
arr
```

```python
arr[:5]
```

```python
arr[2:6:2]
```

```python
arr[::2]
```

```python
arr[1::]
```

```python
arr=np.arange(1,31).reshape(5,6)
arr
```

```python
arr[4]
# Indexing of rows and columns works seperately
# Indexng start with zero
```

```python
arr[2,3]
# To fetch element [row,column]
```

```python
# Slicing
arr[0:2,1:3]
```

```python
arr[0:5,0:4]
```

```python
arr
```

```python
arr[:,2]
#  converts in vector/ list
```

```python
arr[2,:]
```
